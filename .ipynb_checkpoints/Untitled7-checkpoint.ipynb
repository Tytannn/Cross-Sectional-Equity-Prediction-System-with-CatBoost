First we want to select our stock portfolio. We our stocks to be liquid (basically more money moving around via the stock) and diverse. We arbitrarity will choose around 50 stocks for our portfolio. in order to liquidity and diversity criteria, we will calculate the Average Dollar Volume (ADV) per stock per industry. ADV is calculated as taking the average of the product of price and volume traded per given trading period (sorry if the wording is kinda werid). Anyways, We use ADV instead of volume as our liquidity metric because ADV normalizes volume based on price.

```python
df_long["dollar_volume"] = df_long["close"] * df_long["volume"] #dollar volume
#get ADV
avg_dollar_volume = df_long.groupby("symbol")['dollar_volume'].mean().reset_index().rename(columns={"dollar_volume":"avg_dollar_volume"})
universe = sp500.merge(avg_dollar_volume, on='symbol', how = 'inner')#.rename(columns = {"gics sub-industry":"sector"}) #assign ADV to each ticker in a new df

n_per_sector = 5 # 11 sectors * 5 = 55 tickers



selected = universe.sort_values("avg_dollar_volume", ascending = False).groupby('sector').head(n_per_sector) #get n number of tickers per sector
final_tickers = selected["symbol"].tolist()

print(len(final_tickers))
selected["sector"].value_counts()
```

```python
model_df =  df_long[df_long["symbol"].isin(final_tickers)].copy() #make a copy of the subset of the most liquid companies per sector
model_df = model_df.sort_values(["symbol", "date"]).reset_index(drop = True) #sort values by symbol and date
model_df["return_1d"] =  model_df.groupby("symbol")["close"].pct_change() # get the 1 day percent change in price
model_df
```

```python
def add_features(g):
    """Add technical indicators and regime-aware features"""
    g = g.copy()

    # ============================================================
    # ORIGINAL TECHNICAL INDICATORS
    # ============================================================
    # RSI
    g["rsi_14"] = ta.rsi(g["close"], length=14)

    # MACD
    macd_df = ta.macd(g["close"])
    g["macd"] = macd_df["MACD_12_26_9"]

    # Bollinger Bands
    bb = ta.bbands(g["close"], length=20)
    g["bb_pct"] = bb["BBP_20_2.0_2.0"]

    # Volatility
    g["return_1d"] = g["close"].pct_change()
    g["vol_20d"] = g["return_1d"].rolling(20).std()

    # Volume Z-score
    g["volume_z"] = (g["volume"] - g["volume"].rolling(20).mean()) / g["volume"].rolling(20).std()

    # Returns
    g["return_5d"] = g["close"].pct_change(5)
    g["return_10d"] = g["close"].pct_change(10)

    # ADX (Trend Strength)
    adx_df = ta.adx(g["high"], g["low"], g["close"], length=14)
    adx_col = [c for c in adx_df.columns if c.startswith("ADX_")][0]
    g["adx_14"] = adx_df[adx_col]

    # OBV (Volume-Price)
    g["obv"] = ta.obv(g["close"], g["volume"])
    g["obv_z"] = (g["obv"] - g["obv"].rolling(20).mean()) / g["obv"].rolling(20).std()

    # ============================================================
    # REGIME-AWARE FEATURES (NEW)
    # ============================================================

    # 1. MULTIPLE VOLATILITY WINDOWS
    g["vol_5d"] = g["return_1d"].rolling(5).std()
    g["vol_60d"] = g["return_1d"].rolling(60).std()

    # Volatility ratios (regime change signals)
    g["vol_ratio_5_20"] = g["vol_5d"] / g["vol_20d"]
    g["vol_ratio_20_60"] = g["vol_20d"] / g["vol_60d"]

    # 2. MULTIPLE RSI WINDOWS
    g["rsi_5"] = ta.rsi(g["close"], length=5)
    g["rsi_30"] = ta.rsi(g["close"], length=30)

    # 3. MULTIPLE RETURN WINDOWS
    g["return_21d"] = g["close"].pct_change(21)
    g["return_60d"] = g["close"].pct_change(60)

    # 4. MOVING AVERAGES (Trend Regime)
    for ma in [50, 100, 200]:
        g[f"ma_{ma}d"] = g["close"].rolling(ma).mean()
        g[f"price_vs_ma_{ma}"] = g["close"] / g[f"ma_{ma}d"] - 1

    # 5. TREND REGIME INDICATORS
    # Trend strength (ADX is already calculated above)
    g["trend_strength"] = abs(g["return_5d"] / g["vol_20d"])
    g["momentum_stability"] = g["return_5d"].rolling(20).std()

    # 6. REGIME SHIFT DETECTION
    # Rolling correlation between returns and volatility
    g["return_vol_corr"] = g["return_5d"].rolling(20).corr(g["vol_20d"])

    return g
```

```python
# ============================================================
# PREPARE DATA
# ============================================================
# Determine the symbol column name
if 'symbol' in df_long.columns:
    symbol_col = 'symbol'
elif 'ticker' in df_long.columns:
    symbol_col = 'ticker'
elif 'Symbol' in df_long.columns:
    symbol_col = 'Symbol'
elif 'Ticker' in df_long.columns:
    symbol_col = 'Ticker'
else:
    # Find any column that might contain ticker symbols
    possible_cols = [c for c in df_long.columns if 'symbol' in c.lower() or 'ticker' in c.lower()]
    if possible_cols:
        symbol_col = possible_cols[0]
        print(f"Using '{symbol_col}' as symbol column")
    else:
        raise ValueError("No symbol/ticker column found in dataframe")

print(f"Using symbol column: '{symbol_col}'")
print()

# Now create model_df with proper symbol column
model_df = df_long[df_long[symbol_col].isin(final_tickers)].copy()
model_df = model_df.sort_values([symbol_col, "date"]).reset_index(drop=True)
```

```python
# ============================================================
# ADD TECHNICAL FEATURES — Use manual loop instead of groupby
# ============================================================
# Get unique symbols
unique_symbols = model_df[symbol_col].unique()
print(f"Adding features to {len(unique_symbols)} symbols...")

# Create a list to store results
featured_dfs = []

# Process each symbol separately
for symbol in unique_symbols:
    symbol_df = model_df[model_df[symbol_col] == symbol].copy()
    symbol_df = add_features(symbol_df)
    featured_dfs.append(symbol_df)

# Concatenate all results
model_df = pd.concat(featured_dfs, ignore_index=True)

# Ensure symbol column is preserved
if symbol_col not in model_df.columns:
    print(f"ERROR: '{symbol_col}' column was lost!")
    print(f"Available columns: {model_df.columns.tolist()}")
    raise KeyError(f"Column '{symbol_col}' not found")

print(f"Feature addition complete. Shape: {model_df.shape}")
print()
```

```python
# ============================================================
# 1. LOAD + CLEAN MONTHLY COMMODITY DATA
# ============================================================
commodity_raw = pd.read_csv(
    "CMO-Historical-Data-Monthly.csv",
    na_values=["...", "…", ".."],
)

commodity_raw["date"] = pd.to_datetime(
    commodity_raw["Date"].str.replace("M", "-", regex=False), format="%Y-%m"
)
commodity_raw = commodity_raw.drop(columns=["Date"]).sort_values("date")

value_cols = [c for c in commodity_raw.columns if c != "date"]
for c in value_cols:
    commodity_raw[c] = pd.to_numeric(commodity_raw[c], errors="coerce")

commodity_raw[value_cols] = commodity_raw[value_cols].ffill()
# Shift by 2 months: commodity data for month M is typically published near end of month M+1,
# so data for December is not available until late January/early February.
# A 2-month lag ensures no lookahead bias — e.g., January trades only see November data.
commodity_raw["date"] = commodity_raw["date"] + pd.DateOffset(months=1)

print("Commodity NaN rate by column after ffill (top 10):")
print(commodity_raw[value_cols].isna().mean().sort_values(ascending=False).head(10))
print(commodity_raw.shape)
print()
```

```python
# ============================================================
# 2. MERGE ONTO DAILY EQUITY DATA (no lookahead + 2-month publication lag)
# ============================================================
overlap_cols = [c for c in value_cols if c in model_df.columns]
if overlap_cols:
    print(f"Dropping {len(overlap_cols)} already-merged commodity columns before re-merging")
    model_df = model_df.drop(columns=overlap_cols)

model_df["date"] = model_df["date"].astype("datetime64[ns]")
commodity_raw["date"] = commodity_raw["date"].astype("datetime64[ns]")

model_df = model_df.sort_values("date")
commodity_raw = commodity_raw.sort_values("date")

model_df = pd.merge_asof(
    model_df, commodity_raw,
    on="date",
    direction="backward"
)

print("Post-merge NaN rate for commodity features (top 10):")
print(model_df[value_cols].isna().mean().sort_values(ascending=False).head(10))
print()
```

```python
# ============================================================
# 3. CROSS-SECTIONAL REGIME FEATURES (after merge)
# ============================================================
def add_cross_sectional_features(df):
    """Add features that require cross-sectional data"""
    df = df.copy()

    # Relative strength within universe
    df["relative_strength"] = df.groupby("date")["return_5d"].transform(
        lambda x: (x - x.mean()) / x.std()
    )

    # Market breadth (what % of stocks are positive)
    df["market_breadth"] = df.groupby("date")["return_5d"].transform(
        lambda x: (x > 0).mean()
    )

    # Volatility percentile within universe
    df["vol_percentile"] = df.groupby("date")["vol_20d"].transform(
        lambda x: x.rank(pct=True)
    )

    # Cross-sectional momentum dispersion
    df["momentum_dispersion"] = df.groupby("date")["return_5d"].transform(
        lambda x: x.std()
    )

    return df

model_df = add_cross_sectional_features(model_df)
```

```python
# ============================================================
# 4. LABEL: 5-day forward return, cross-sectional rank vs. universe on that date
# ============================================================
FORWARD_DAYS = 5

# CRITICAL FIX: ensure each symbol's rows are chronologically ordered
# before shifting within the symbol group.
model_df = model_df.sort_values([symbol_col, "date"]).reset_index(drop=True)

# Safety: drop duplicate (symbol, date) rows if any exist
dups = model_df.duplicated(subset=[symbol_col, "date"]).sum()
if dups > 0:
    print(f"Warning: dropping {dups} duplicate ({symbol_col}, date) rows")
    model_df = model_df.drop_duplicates(subset=[symbol_col, "date"]).reset_index(drop=True)

model_df["fwd_return_5d"] = (
    model_df.groupby(symbol_col)["close"].shift(-FORWARD_DAYS) / model_df["close"] - 1
)
model_df["fwd_return_rank"] = model_df.groupby("date")["fwd_return_5d"].rank(pct=True)
model_df["label"] = (model_df["fwd_return_rank"] > 0.5).astype(int)

print("Correlation check (should not be high):")
# Check if return_5d exists, if not, use return_1d
if "return_5d" in model_df.columns:
    print(model_df[["return_5d", "fwd_return_5d"]].corr())
else:
    print("Note: return_5d not found, skipping correlation check")
print()
```

```python
# ============================================================
# 5. CONFIG
# ============================================================
# Include all new regime-aware features
base_features = [
    # Original features
    "rsi_14", "macd", "bb_pct", "vol_20d", "volume_z", "return_5d", "return_10d",

    # New technical features
    "adx_14", "obv_z",

    # Multi-window features
    "rsi_5", "rsi_30",
    "vol_5d", "vol_60d",
    "vol_ratio_5_20", "vol_ratio_20_60",
    "return_1d", "return_21d", "return_60d",

    # Trend regime
    "price_vs_ma_50", "price_vs_ma_100", "price_vs_ma_200",
    "trend_strength", "momentum_stability",

    # Regime detection
    "return_vol_corr",

    # Cross-sectional features
    "relative_strength", "market_breadth",
    "vol_percentile", "momentum_dispersion"
]

# Filter to only include features that actually exist in the dataframe
existing_features = [f for f in base_features if f in model_df.columns]
missing_features = [f for f in base_features if f not in model_df.columns]

if missing_features:
    print(f"Warning: {len(missing_features)} features are missing from dataframe:")
    print(f"  {missing_features[:10]}")  # Show first 10 missing

base_features = existing_features
print(f"Using {len(base_features)} base features")

commodity_nan_rate = commodity_raw[value_cols].isna().mean()
dropped_commodity_cols = commodity_nan_rate[commodity_nan_rate >= 0.5].index.tolist()
commodity_features = commodity_nan_rate[commodity_nan_rate < 0.5].index.tolist()

print(f"Dropping {len(dropped_commodity_cols)} commodity columns with >=50% missing: {dropped_commodity_cols}")
print(f"Keeping {len(commodity_features)} commodity features")
print()

# Allow separate feature sets for ablation studies
USE_COMMODITIES = True  # Set to False for technical-only baseline
if USE_COMMODITIES:
    feature_cols = commodity_features + base_features
    print("Using technical indicators + commodities")
else:
    feature_cols = base_features.copy()
    print("Using technical indicators only (ablation mode)")

print(f"\nTotal features: {len(feature_cols)}")
print(f"Base technical features: {len(base_features)}")
print(f"Commodity features: {len(commodity_features) if USE_COMMODITIES else 0}")

LABEL_HORIZON_DAYS = 5
TRAIN_WINDOW_DAYS = 500
RETRAIN_EVERY_DAYS = 20
MIN_TRAIN_DAYS = 250
MIN_SYMBOLS_PER_DATE = 10
VALIDATION_SPLIT = 0.2  # Fraction of training data for early stopping validation
```

```python
# ============================================================
# 6. FEATURE NORMALIZATION
# ============================================================
def cross_sectional_zscore(df, cols):
    df = df.copy()
    zscored = {}
    for col in cols:
        if col in df.columns:
            grp = df.groupby("date")[col]
            zscored[col] = (df[col] - grp.transform("mean")) / grp.transform("std").replace(0, np.nan)
    return df.assign(**zscored)

# Z-score normalize all features (including new ones)
all_features_to_normalize = base_features.copy()
if USE_COMMODITIES:
    all_features_to_normalize.extend(commodity_features)

# Only normalize features that exist
all_features_to_normalize = [f for f in all_features_to_normalize if f in model_df.columns]
model_df = cross_sectional_zscore(model_df, all_features_to_normalize)

# CRITICAL FIX: commodity values are identical for all symbols on a given date.
# Rolling(60) on row-sorted data counts rows (symbols), not dates.
# We must compute the rolling window on the unique date series, then broadcast back.
if USE_COMMODITIES:
    for col in commodity_features:
        if col in model_df.columns:
            daily_vals = model_df.groupby("date")[col].first().sort_index()
            roll_mean = daily_vals.rolling(60, min_periods=20).mean()
            roll_std = daily_vals.rolling(60, min_periods=20).std().replace(0, np.nan)
            daily_z = (daily_vals - roll_mean) / roll_std
            model_df[col] = model_df["date"].map(daily_z)
```

```python
# ============================================================
# 7. BREADTH FILTER
# ============================================================
symbols_per_date = model_df.groupby("date")[symbol_col].transform("nunique")
model_df = model_df[symbols_per_date >= MIN_SYMBOLS_PER_DATE].copy()

global_baseline = model_df["label"].value_counts(normalize=True).max()
print(f"Global naive baseline (always predict majority class): {global_baseline:.3f}")
print()

print("NaN rate per feature (informational only -- not used to filter rows):")
nan_rates = model_df[feature_cols].isna().mean()
print(nan_rates.sort_values(ascending=False).head(20))
print()

# ============================================================
# 7b. PRE-LOOP DIAGNOSTIC: fwd_return_5d validity per date
# ============================================================
print("=" * 60)
print("PRE-LOOP DIAGNOSTIC: fwd_return_5d valid count per date")
overall_valid = model_df.groupby("date")["fwd_return_5d"].apply(lambda s: s.notna().sum())
print(f"  min={overall_valid.min()}, max={overall_valid.max()}, mean={overall_valid.mean():.1f}")
print(f"  dates with <=5 valid fwd_return_5d: {(overall_valid <= 5).sum()} of {len(overall_valid)}")
print("=" * 60)
print()
```

```python
# ============================================================
# 8. PERIODIC-RETRAIN WALK-FORWARD, CROSS-SECTIONAL
# ============================================================
dates = pd.Series(model_df["date"].unique()).sort_values().reset_index(drop=True)
n_dates = len(dates)

fold_results = []
fold_models = []
all_preds = []

period_start_idx = MIN_TRAIN_DAYS
fold = 0
skipped_empty = 0
skipped_single_class = 0

while period_start_idx < n_dates:
    period_end_idx = min(period_start_idx + RETRAIN_EVERY_DAYS, n_dates)

    test_dates = dates.iloc[period_start_idx:period_end_idx]
    test_start = test_dates.min()
    test_end = test_dates.max()

    train_window_start_idx = max(0, period_start_idx - TRAIN_WINDOW_DAYS)
    train_dates_window = dates.iloc[train_window_start_idx:period_start_idx]

    purge_cutoff = test_start - pd.Timedelta(days=LABEL_HORIZON_DAYS)
    train_dates_final = train_dates_window[train_dates_window < purge_cutoff]

    fold_train = model_df[model_df["date"].isin(train_dates_final)].dropna(subset=["label"])
    fold_test = model_df[model_df["date"].isin(test_dates)].dropna(subset=["label"])

    if len(fold_train) == 0 or len(fold_test) == 0:
        skipped_empty += 1
        period_start_idx = period_end_idx
        continue

    X_train, y_train = fold_train[feature_cols], fold_train["label"]
    X_test, y_test = fold_test[feature_cols], fold_test["label"]

    if y_train.nunique() < 2 or y_test.nunique() < 2:
        skipped_single_class += 1
        period_start_idx = period_end_idx
        continue

    fold_baseline = y_test.value_counts(normalize=True).max()

    # OPTIMIZED: Split training data into train/validation for early stopping
    X_tr, X_val, y_tr, y_val = train_test_split(
        X_train, y_train,
        test_size=VALIDATION_SPLIT,
        random_state=42,
        stratify=y_train
    )

    # OPTIMIZED: CatBoost with faster settings
    cat = CatBoostClassifier(
        random_seed=42,
        verbose=False,
        iterations=80,              # Limit total iterations
        early_stopping_rounds=20,   # Fewer patience rounds
        depth=4,                    # Shallower trees for speed
        l2_leaf_reg=5,
        random_strength=1,
        subsample=0.8,              # Use 80% of data per tree
        colsample_bylevel=0.7,      # Use 70% of features per level
        min_data_in_leaf=20,        # Prevent overfitting
        border_count=32,            # Fewer split bins for speed
    )
    cat.fit(X_tr, y_tr, eval_set=(X_val, y_val))

    preds = cat.predict(X_test)
    probs = cat.predict_proba(X_test)[:, 1]

    acc = accuracy_score(y_test, preds)
    auc = roc_auc_score(y_test, probs)
    prob_std = probs.std()

    n_purged = len(train_dates_window) - len(train_dates_final)

    print(f"Period {fold}: train {fold_train['date'].min().date()} to {fold_train['date'].max().date()} "
          f"| test {test_start.date()} to {test_end.date()} "
          f"— acc: {acc:.3f}, local_baseline: {fold_baseline:.3f}, auc: {auc:.3f}, "
          f"prob_std: {prob_std:.5f}, "
          f"n_test: {len(y_test)}, n_val: {len(y_val)}, dates_purged: {n_purged}")

    fold_results.append({
        "fold": fold,
        "accuracy": acc,
        "local_baseline": fold_baseline,
        "auc": auc,
        "prob_std": prob_std,
        "n_test": len(y_test),
        "n_train": len(y_tr),
        "n_val": len(y_val),
        "dates_purged": n_purged,
        "train_start": fold_train["date"].min(),
        "train_end": fold_train["date"].max(),
        "test_start": test_start,
        "test_end": test_end
    })
    fold_models.append(cat)

    all_preds.append(pd.DataFrame({
        "date": fold_test["date"].values,
        symbol_col: fold_test[symbol_col].values,
        "y_true": y_test.values,
        "y_pred": preds,
        "y_prob": probs,
        "fwd_return_5d": fold_test["fwd_return_5d"].values
    }))

    period_start_idx = period_end_idx
    fold += 1

print()
print(f"Periods skipped (empty train/test): {skipped_empty}")
print(f"Periods skipped (single class): {skipped_single_class}")
print(f"Periods populated: {fold} out of {n_dates // RETRAIN_EVERY_DAYS} possible")
print()

results_df = pd.DataFrame(fold_results)

if len(results_df) == 0:
    print("No folds were populated -- check the NaN-rate printout above.")
else:
    results_df["beats_baseline"] = results_df["accuracy"] > results_df["local_baseline"]

    pooled = pd.concat(all_preds, ignore_index=True)
    pooled_acc = accuracy_score(pooled["y_true"], pooled["y_pred"])
    pooled_auc = roc_auc_score(pooled["y_true"], pooled["y_prob"])
    pooled_baseline = pooled["y_true"].value_counts(normalize=True).max()

    print(f"Pooled out-of-sample accuracy: {pooled_acc:.3f} (baseline: {pooled_baseline:.3f})")
    print(f"Pooled out-of-sample AUC: {pooled_auc:.3f}")
    print()

    # Long-short portfolio backtest
    print("=" * 60)
    print("LONG-SHORT PORTFOLIO (TOP VS BOTTOM QUINTILE)")
    pooled["prob_quintile"] = pooled.groupby("date")["y_prob"].transform(
        lambda x: pd.qcut(x, 5, labels=False, duplicates='drop')
    )

    # Long top quintile, short bottom quintile
    top_quintile = pooled[pooled["prob_quintile"] == 4].groupby("date")["fwd_return_5d"].mean()
    bottom_quintile = pooled[pooled["prob_quintile"] == 0].groupby("date")["fwd_return_5d"].mean()

    # Align dates
    common_dates = top_quintile.index.intersection(bottom_quintile.index)
    long_short = top_quintile[common_dates] - bottom_quintile[common_dates]

    if len(long_short) > 0:
        print(f"  Mean daily spread: {long_short.mean():.4%}")
        print(f"  Std daily spread: {long_short.std():.4%}")
        print(f"  Sharpe ratio (daily): {long_short.mean() / long_short.std():.3f}")
        print(f"  Annualized Sharpe: {long_short.mean() / long_short.std() * np.sqrt(252):.3f}")
        print(f"  Fraction positive: {(long_short > 0).mean():.2%}")
    else:
        print("  Not enough data for long-short analysis")
    print("=" * 60)
    print()

    # Performance over time
    print("=" * 60)
    print("PERFORMANCE BY TIME PERIOD")
    results_df["year"] = pd.to_datetime(results_df["test_start"]).dt.year
    yearly_perf = results_df.groupby("year").agg({
        "accuracy": "mean",
        "auc": "mean",
        "beats_baseline": "mean"
    }).round(4)
    print(yearly_perf)
    print("=" * 60)

print(results_df)
```

```python
# ============================================================
# 9. FEATURE IMPORTANCE & PERMUTATION IMPORTANCE
# ============================================================
if len(fold_models) > 0 and len(feature_cols) > 0:
    print("=" * 60)
    print("FEATURE IMPORTANCE ANALYSIS")
    print("=" * 60)

    # --- A) CatBoost built-in feature importance (from last model) ---
    last_model = fold_models[-1]
    catboost_imp = pd.DataFrame({
        "feature": feature_cols,
        "importance": last_model.get_feature_importance()
    }).sort_values("importance", ascending=False)

    # --- B) Permutation importance using last fold's test data ---
    # Use all_preds (date + symbol) to locate the test rows in model_df — avoids date-dtype issues
    last_preds = all_preds[-1].copy()
    last_preds["date"] = pd.to_datetime(last_preds["date"])

    # Merge on date + symbol to get feature values (model_df already has normalized features)
    X_last_test = last_preds[["date", symbol_col]].merge(
        model_df[["date", symbol_col] + feature_cols],
        on=["date", symbol_col],
        how="inner"
    )[feature_cols].dropna()

    y_last_test = last_preds.loc[X_last_test.index, "y_true"]

    # Keep it manageable: sample up to 5000 rows for speed
    if len(X_last_test) > 5000:
        sample_idx = X_last_test.sample(5000, random_state=42).index
        X_sample = X_last_test.loc[sample_idx]
        y_sample = y_last_test.loc[sample_idx]
    else:
        X_sample = X_last_test
        y_sample = y_last_test

    print(f"Computing permutation importance on {len(X_sample)} samples (last fold test set)...")

    if len(X_sample) == 0:
        print("  WARNING: 0 test samples available — skipping permutation importance.")
    else:
        perm_result = permutation_importance(
            last_model, X_sample, y_sample,
            n_repeats=10,
            random_state=42,
            scoring="roc_auc",
            n_jobs=-1
        )

        perm_imp = pd.DataFrame({
            "feature": feature_cols,
            "importance_mean": perm_result["importances_mean"],
            "importance_std": perm_result["importances_std"]
        }).sort_values("importance_mean", ascending=False)

    # --- C) Plot top 20 features from both methods ---
    fig, axes = plt.subplots(1, 2, figsize=(16, 10))

    # CatBoost importance (top 20)
    top20_cb = catboost_imp.head(20)
    axes[0].barh(range(len(top20_cb)), top20_cb["importance"].values[::-1], color="steelblue")
    axes[0].set_yticks(range(len(top20_cb)))
    axes[0].set_yticklabels(top20_cb["feature"].values[::-1])
    axes[0].set_xlabel("Importance")
    axes[0].set_title("CatBoost Built-in Feature Importance (Top 20)")
    axes[0].invert_yaxis()

    # Permutation importance (top 20) — only if we have results
    if len(X_sample) > 0:
        top20_perm = perm_imp.head(20)
        axes[1].barh(
            range(len(top20_perm)),
            top20_perm["importance_mean"].values[::-1],
            xerr=top20_perm["importance_std"].values[::-1],
            color="darkorange", capsize=3
        )
        axes[1].set_yticks(range(len(top20_perm)))
        axes[1].set_yticklabels(top20_perm["feature"].values[::-1])
        axes[1].set_xlabel("Mean AUC Decrease")
        axes[1].set_title("Permutation Importance (Top 20)")
        axes[1].invert_yaxis()
    else:
        axes[1].text(0.5, 0.5, "No permutation data", ha="center", va="center", transform=axes[1].transAxes)
        axes[1].set_title("Permutation Importance (skipped)")

    plt.tight_layout()
    plt.show()

    # --- D) Breakdown: technical vs commodity importance ---
    print("--- Feature Category Breakdown ---")
    tech_mask = catboost_imp["feature"].isin(base_features)
    comm_mask = catboost_imp["feature"].isin(commodity_features)

    tech_total = catboost_imp.loc[tech_mask, "importance"].sum()
    comm_total = catboost_imp.loc[comm_mask, "importance"].sum()
    total_imp = tech_total + comm_total

    if total_imp > 0:
        print(f"  Technical indicators: {tech_total:.2f} ({tech_total/total_imp*100:.1f}%)")
        print(f"  Commodity features:   {comm_total:.2f} ({comm_total/total_imp*100:.1f}%)")

    # Top commodity features
    top_commodities = catboost_imp[comm_mask].head(10)
    if len(top_commodities) > 0:
        print(f"Top commodity features:")
        for _, row in top_commodities.iterrows():
            print(f"    {row['feature']}: {row['importance']:.4f}")

    # --- E) Print full ranking ---
    if len(X_sample) > 0:
        print("--- Full Permutation Importance Ranking ---")
        for i, (_, row) in enumerate(perm_imp.iterrows()):
            cat = "COMM" if row["feature"] in commodity_features else "TECH"
            print(f"  {i+1:2d}. [{cat}] {row['feature']:<35s} {row['importance_mean']:+.6f} ± {row['importance_std']:.6f}")
            if i >= 29:
                break
    else:
        print("--- CatBoost Feature Importance (fallback) ---")
        for i, (_, row) in enumerate(catboost_imp.iterrows()):
            cat = "COMM" if row["feature"] in commodity_features else "TECH"
            print(f"  {i+1:2d}. [{cat}] {row['feature']:<35s} {row['importance']:.4f}")
            if i >= 29:
                break

else:
    print("No models trained — cannot compute feature importance.")
```
